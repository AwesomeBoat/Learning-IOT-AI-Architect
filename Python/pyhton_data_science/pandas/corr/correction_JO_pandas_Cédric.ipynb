```python
import pandas as pd
```

```python
data = pd.read_csv("data/Exercices_Pandas_athlete_events.csv")
```

```python
"""22. Quels sont les joueurs ayant gagné des médailles d’Or en badminton simple au cours des
années. Trier le résultat par ordre alphabétique sur le nom des joueurs."""

cond_gold = data["Medal"] == "Gold"
cond_badminton_simple = (data["Event"].str.contains("single",case=False)) & (data["Sport"]=="Badminton")

print((cond_gold&cond_badminton_simple).sum())

(
    data
    .loc[cond_gold&cond_badminton_simple,:]
    .sort_values(by=["Name"],ascending=[True])
)
```

```python
"""32) Quels étaient les pays les plus efficaces durant la guerre froide de 1952 à 1988 (i.e. le pays
avec le plus de médailles par nombre d’athlètes envoyés)"""

cond_year= data["Year"].between(1952,1988, inclusive="both")
GB = (
    data
    .loc[cond_year,:]
    .groupby(["NOC"],as_index=0)
    .agg(count_medal = ("Medal","count"), count_athl=("ID","count"))
    .sort_values(by="count_medal",ascending=False)
)

GB["ratio"] = GB["count_medal"]/GB["count_athl"]
GB
```

```python
GB2 = (
    data
    .drop_duplicates(subset=["NOC","Medal","Year","Event"])
    .loc[cond_year,:]
    .groupby(["NOC"],as_index=0)
    .agg(count_medal = ("Medal","count"), count_athl=("ID","count"))
    .sort_values(by="count_medal",ascending=False)
)
GB2["ratio"] = GB2["count_medal"]/GB2["count_athl"]
GB2.sort_values(by="ratio",ascending=0)
```
