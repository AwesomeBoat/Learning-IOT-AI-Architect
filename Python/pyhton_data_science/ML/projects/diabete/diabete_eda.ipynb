```python
#4.3 Analyser la distribution pour chaque variable

for col in df.columns:
    if col == 'Unnamed: 0' :
        continue
    elif col == "Education":
        plt.figure(figsize=(4,6))
        plt.hist(df["Education"].dropna(), bins=30, rwidth=0.6)
        plt.xticks(fontsize=7)
        plt.xlabel(col)
        plt.ylabel("frequence")
        plt.title(f"Distribution de {col}")
        plt.show()
        continue
    plt.figure(figsize=(4,6))
    plt.hist(df[col])
    plt.xlabel(col)
    plt.ylabel("frequence")
    plt.title(f"Distribution de {col}")
    plt.show()
```
