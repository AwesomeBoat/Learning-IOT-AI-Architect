```python
dict_donnees["math"][:10] ## 10 premières lignes de la colonne 'math'

# =================== PARTIE 1 ===================

# Effectif gender
m_effectif = len([e for e in dict_donnees["gender"] if e =="M"])
f_effectif = len([e for e in dict_donnees["gender"] if e =="F"])

gender_effectif = {"M":m_effectif,
                   "F":f_effectif}
print(f"Effectif de gender : {gender_effectif}")


# Frequences gender

ntot = gender_effectif["M"]+gender_effectif["F"]

frequences_gender = {
    "M":gender_effectif["M"]/ntot,
    "F":gender_effectif["F"]/ntot
}

print(f"frequence de gender : {frequences_gender}")


# Effectif group


lang_effectif = len([e for e in dict_donnees["group"] if e =="lang"])
science_effectif = len([e for e   in dict_donnees["group"] if e =="science"])
group_effectif = {
    "lang":lang_effectif,
    "science":science_effectif
}
print(f"Effectif de group : {group_effectif}")


# frequence de group

frequences_group = {
    "lang":group_effectif["lang"]/ntot,
    "science":group_effectif["science"]/ntot
}

print(f"Frequence de group : {frequences_group}")

""" 
Frequences conjointes (Combien de personne du genre "M" appartient au groupe lang,
                           Combien de personne du genre "F" appartient au groupe science,
                           Combien de personne du genre "M" appartient au groupe lang,
                           Combien de personne du genre "F" appartient au groupe science            )
"""



# Création var effectif conjoint gender_group

M_lang_effectif = 0
F_lang_effectif = 0
M_science_effectif = 0
F_science_effectif = 0

# Itération du dictionnaire
for i in range (ntot):
    if dict_donnees["gender"][i] =="M" and dict_donnees["group"][i] =="lang":
        M_lang_effectif += 1
    elif dict_donnees["gender"][i] =="F" and dict_donnees["group"][i] =="lang":
        F_lang_effectif += 1
    elif dict_donnees["gender"][i] =="F" and dict_donnees["group"][i] =="science":
        F_science_effectif += 1
    elif dict_donnees["gender"][i] =="M" and dict_donnees["group"][i] =="science":
        M_science_effectif += 1

effectif_group_gender = {
    "M_lang":M_lang_effectif,
    "F_lang":F_lang_effectif,
    "M_science":M_science_effectif,
    "F_science":F_science_effectif
}

frequences_group_gender = {
    "M_lang":M_lang_effectif/ntot,
    "F_lang":F_lang_effectif/ntot,
    "M_science":M_science_effectif/ntot,
    "F_science":F_science_effectif/ntot
}

print(f"Effectif group/gender :  {effectif_group_gender}")
print(f"frequences_conjointes_group_gender : {frequences_group_gender}")


"""   
Fréquences conditionnelles: Par groupe, proportionnellement, Combien y'a t'il de femme en langue % total, d'homme en langue %tot, ...
"""

# Parmis les étudiants en langues, combien sont des femmes, combien sont des hommes ? 
ntot_langue = group_effectif["lang"] 
ntot_science = group_effectif["science"]

# mon % de femme en lang = nbre_femmes en lange / ntot langues *100
pourcentage_femme_langue = (frequences_group_gender["F_lang"]/ntot_langue)*100
pourcentage_homme_langue = (frequences_group_gender["M_lang"]/ntot_langue)*100

pourcentage_femme_science = (frequences_group_gender["F_science"]/ntot_science)*100
pourcentage_homme_science = (frequences_group_gender["M_science"]/ntot_science)*100

freq_cond = {
    "ML" : pourcentage_homme_langue,
    "FL" : pourcentage_femme_langue,
    "MS" : pourcentage_homme_science,
    "FS" : pourcentage_femme_science
}

print("frequence conditionnelle : {freq_cond}")
```

```python
import math
""" 

=============== PARTIE 2 ===============

à calculer en **algorithme python** (natif, depuis zéro) (ou dans Excel si vous n'avez pas vu python):
- calculer les mesures de tendance centrale (appropriés) pour chaque colonne
- calculer les mesures de dispersions (appropriés) pour chaque colonne
- calculer les corrélations (de pearson) pour les colonnes appropriées

"""

# ===== 1. calculer les mesures de tendance centrale (appropriés) pour chaque colonne =====

def tendance_centrale(dict,colonne,type):

    match type:
        case "moyenne_arithm":
            return sum([e for e in dict[colonne]])/len(dict[colonne])
        
        case "medianne":
            sorted_list = sorted(dict[colonne])
            longueur_colonne = len(dict[colonne])
            if longueur_colonne%2==0: # cas pair

                middle_index_left = longueur_colonne//2-1
                middle_index_right = (longueur_colonne//2)
                return (sorted_list[middle_index_left]+dict[colonne][middle_index_right])/2

            
            else : # cas impair
                middle_index = longueur_colonne//2
                return dict[colonne][middle_index]

        case "mode": 
            dict_effectif = {}

            for el in dict[colonne]:
                if el not in dict_effectif:

                    dict_effectif[el] = 1
                else:
                    dict_effectif[el] += 1
            return  max(dict_effectif, key=dict_effectif.get)
        


# ===== 2. calculer les mesures de dispersions (appropriés) pour chaque colonne =====

def mesure_dispertion(dict,colonne,type):

    match type:

        case "amplitude": # L'amplitude est applicable quand la médiane est applicable => Distance entre le max et le min : valeur max - valeur min = amplitude
            return max(dict[colonne]) - min(dict[colonne])
        
        case "deviation_absolue_moyenne": # La déviation abs. moyenne est applicable quand la moyenne est applciable => La moyenne des valeurs absolues de l'écart entre la valeur et la moyenne 
            moyenne = tendance_centrale(dict,colonne,"moyenne_arithm")
            ecart_valeur_moyenne = [abs(el-moyenne) for el in dict[colonne]]
            return sum(ecart_valeur_moyenne)/len(ecart_valeur_moyenne)
        

        case "variance": # La variance / ecart-type s'applique quand la moyenne est applicable => la  moyenne des carré des valeurs de m'écart entre la valeur et la moyenne
            moyenne = tendance_centrale(dict,colonne,"moyenne_arithm")
            ecart_valeur_moyenne = [(el-moyenne)**2 for el in dict[colonne]]
            return sum(ecart_valeur_moyenne)/len(ecart_valeur_moyenne)
    

        case "ecart_type": # L'écart type s'applique quand la variance estréalisable donc moyenne => 
            return math.sqrt(mesure_dispertion(dict, colonne, "variance"))
        
        case "quartile": # La quartile est applicable quand la médiane est applicable
            # Trier la liste
            sorted_data = sorted(dict[colonne])
            list_size = len(sorted_data)
            medianne = tendance_centrale(dict,colonne,"medianne")
            return {
                "Q1":sorted_data[list_size//4],
                "Q2":medianne,
                "Q3":sorted_data[(list_size//4)*3],
                "Ecart":(sorted_data[(list_size//4)*3])-(sorted_data[list_size//4])
                
            }
           
            

            


# ====== 3. calculer les corrélations (de pearson) pour les colonnes appropriées ======

def correlation(dict_donnees,col1,col2) :
    # On a besoin de 2 listes pour faire la corrélation, je considère donc que on peut entrer une liste dans le paramètre colonne de ma fonction
    # Faire la moyenne de chacune des colonnes
    first_moyenne = tendance_centrale(dict,col1,"moyenne_arithm")
    second_moyenne = tendance_centrale(dict,col2,"moyenne_arithm")



# ==== VAR TENDANCE CENTRALE ====

# Moyenne arithmétique (discrete, continue)
moyenne_arithm_math = tendance_centrale(dict_donnees,"math","moyenne_arithm")
moyenne_arithm_eng = tendance_centrale(dict_donnees,"eng","moyenne_arithm")

# Medianne (ordinale, numériques)
mediane_math = tendance_centrale(dict_donnees,"math","medianne")
mediane_eng = tendance_centrale(dict_donnees,"eng","medianne")
# Mode (Nominale, ordinale, discrete)
mode_gender = tendance_centrale(dict_donnees,"gender","mode")
mode_group = tendance_centrale(dict_donnees,"group","mode" )


# ==== VAR DISPERSION ====

# Amplitude
math_amplitude = mesure_dispertion(dict_donnees, "math", "amplitude")
eng_amplitude = mesure_dispertion(dict_donnees, "eng", "amplitude")

# Deviation_abs_moyenne
math_deviation = mesure_dispertion(dict_donnees,"math","deviation_absolue_moyenne")
eng_deviation = mesure_dispertion(dict_donnees,"eng","deviation_absolue_moyenne")

# Variance
math_variance = mesure_dispertion(dict_donnees,"math","variance")
eng_variance = mesure_dispertion(dict_donnees,"eng","variance")

# Ecart-type
math_ecart_type = mesure_dispertion(dict_donnees,"math","ecart_type")
eng_ecart_type = mesure_dispertion(dict_donnees,"eng","ecart_type")

# Quartiles
math_quartile = mesure_dispertion(dict_donnees,'math',"quartile")
eng_quartile = mesure_dispertion(dict_donnees,"eng","quartile")

# ===== PRINT =====

print("==========TENDANCE CENTRALE==========")
# Tendance centrale pour math
print(f"math moyenne  {moyenne_arithm_math} ")
print(f"math medianne  {moyenne_arithm_math} ")

# Tendance centrale pour anglais :
print(f"eng moyenne  {moyenne_arithm_eng} ")
print(f"eng: medianne  {moyenne_arithm_eng} ")

# Tendance centrale pour gender
print(f"gender mode {mode_gender}")

# Tendance centrale pour group
print(f"group mode {mode_group}")



print("\n==========DISPERSION==========")
print(f"Amplitude math : {math_amplitude}")
print(f"Amplitude eng : {eng_amplitude}")
print(50*"-")
print(f"Deviation abs. moyenne math : {math_deviation}")
print(f"Deviation abs. moyenne eng : {eng_deviation}")
print(50*"-")
print(f"Variance math : {math_variance}")
print(f"Variance eng : {eng_variance}")
print(50*"-")
print(f"ecart-type math : {math_ecart_type}")
print(f"ecart-type eng : {eng_ecart_type}")
print(50*"-")
print(f"Quartile math : {math_quartile}")
print(f"Quartile eng : {eng_quartile}")
print(50*"-")

```
