## Effectifs, fréquences et distributions
- applicables à toutes les catégories (et indirectement aux numériques)
- EN: **frequencies, relative frequencies** and distributions
- l'idée ici, c'est avoir un aperçu des valeurs possibles et à quelle point ces valeurs sont répétées
    - on ne résume plus avec une ou deux valeurs 
    - on fait des "group by" = tableau de contingence = table pivot = tableau croisé dynamique
    - EN: "group by", contingency table, pivot table 

- Pour des données numériques (en particulier les continues), il est nécessaire de créer des compartiments (=un ensemble d'intervalles) (EN: bins).
    - Ensuite, on pourra regrouper toutes les lignes qui partagent le même compartiment ensemble.
    - le compartiment $[a,b) = [a,b[$ : toutes les valeurs entre $a$ (inclus) et $b$ (exclus)


#### 0. Distribution 
* = parmi les valeurs possibles, comment sont-elles réparties ? Dans quelles proportions ? 

#### 1. (Distribution des) effectifs $F(X)$

* = nombre d’apparitions d’une modalité pour une variable donnée $F(x_i)$ avec $x_i \in X$
* = simplement : "groupby + count"

#### 3. (Distribution des) fréquences $f(X)$ 
* = fréquence d’apparition d’une modalité pour une variable donnée $f(x_i) = F(x_i)/N_\text{tot}$
* avec $N_\text{tot} = \sum_i F(x_i)$

#### 4. (Distribution des) effectifs cumulés 
* = somme cummulé des effectifs 

<img src="repos/effectifs.png" alt="drawing" style="width:400px;"/>



```python

df = pd.DataFrame({"sex":["F","M","F","M","F","F","M"], "age":[17,18,19,21,25,28,35]})
df["age_bin"] = pd.cut(df["age"],bins=[15,20,30,40],right=False) ## creation de compartiment
df["age_bin_size"] = (pd.IntervalIndex(df["age_bin"])).length
print("le dataset avec des compartiments pour la colonne age")
display(df)

print("-"*50)

gb_sex = df.groupby(["sex"],as_index=False).agg(FREQ=("sex","count"))
gb_sex["rel_freq"] = gb_sex["FREQ"]/gb_sex["FREQ"].sum()

print("Effectifs et fréquence de la colonne 'sex' ")
display(gb_sex)

print("-"*50)

gb_age = (df
          .groupby(["age_bin"],as_index=False, observed=False)
          .agg(FREQ=("age_bin","count"))
         )
gb_age["rel_freq"] = gb_age["FREQ"]/gb_age["FREQ"].sum()
gb_age["rel_cummul"] = gb_age["rel_freq"].cumsum()
"""
densité:
- la fréquence (relative frequency) divisé par la taille du compartiment
- pour que la superficie total = 1
"""
gb_age["density"] = gb_age["rel_freq"]/(pd.IntervalIndex(gb_age["age_bin"])).length 
print("Effectifs et fréquence de la colonne 'age' (avec compartiment) ")
display(gb_age)
```


#### 4. Fréquences conjointes $f(X,Y)$

= pour chaque categorie venant des combinaisons (conjointe) de $(X,Y) = \{(x_1,y_1), (x_2,y_1), (x_1,y_2), ... , (x_I, y_K)\}$ quelles sont les fréquences $f(x_i,y_k)$.
- EN: joint relative frequency
  
#### 5. Distribution marginale $f(\star)$ 
= somme des effectifs pour une colonne ou ligne particulière
- $ 1 = f(\cdot) = f(\star) = \sum_{i} f(x_i)$
- $f(x_i) = f(x_i,\star) = \sum_k f(x_i,y_k)$
- $f(y_k) = f(\star,y_k) = \sum_i f(x_i,y_k)$



```python
gb_joint = df.groupby(["sex","age_bin"],as_index=False,observed=True).agg(FREQ=("age_bin","count"))
print("""F(X,Y) effectifs conjoints""")
display(gb_joint)


pivot_joint_FREQ = df.pivot_table(
    index=["sex"]
    ,columns=["age_bin"]
    ,values = "age"
    ,aggfunc= "count"
    ,observed=False
    ,margins=True
)

display(pivot_joint_FREQ)

print("-"*50)
pivot_joint_rel = pivot_joint_FREQ/pivot_joint_FREQ.at["All","All"]
print("""f(X,Y) fréquences conjointes""")
display(pivot_joint_rel)
```

6. Frequences conditionnelles $f(X|Y)$ = pour les données d'une catégorie spécifique $y_k \in Y$, quelles sont les fréquences $f(X|y_k)$.
    - EN: conditionnal relative frequency
    - à noter que:
        - $f(x_i|y_k) = \frac{f(x_i,y_k)}{f(y_k)}$
        - $\sum_k f(x_i|y_k)\cdot f(y_k) = f(x_i)$
    
<img src="repos/effectifs_2.png" alt="drawing" style="width:800px;"/>

```python
pivot_cond_age = pivot_joint_FREQ.divide(pivot_joint_FREQ.loc["All",:],axis=1)
print("f(X|age) fréquence conditionnel sur 'age'")
display(pivot_cond_age)

print("-"*50)

pivot_cond_sex = pivot_joint_FREQ.divide(pivot_joint_FREQ.loc[:,"All"],axis=0)
print("f(X|sex) fréquence conditionnel sur 'sex'")
display(pivot_cond_sex)
```
