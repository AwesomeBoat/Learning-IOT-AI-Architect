```python
import pandas as pd

df = pd.read_csv("treatment.csv")
df["recovery_in_3_days"] = df["recovery_in_3_days"].map({0:"No",1:"Yes"})
df
```

```python
gb= (
    df
    .groupby(["treatment_category","recovery_in_3_days"],as_index=False)
    .agg(count=("treatment_category","count"))
)
gb
```

```python
gb.pivot_table(index=["treatment_category"]
               ,columns=["recovery_in_3_days"]
               ,values='count'
               ,aggfunc="sum"
               , margins=True
              )
```
