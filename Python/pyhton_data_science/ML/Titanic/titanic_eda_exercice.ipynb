## 3. Visualisations univariées

Réalisez des graphiques simples pour comprendre la distribution des variables :

1. Histogramme des âges (`sns.histplot`).
2. Distribution du prix des billets (`sns.histplot`, en log si nécessaire).
3. Répartition des passagers par classe (`sns.countplot`).
4. Répartition du nombre de survivants (`sns.countplot`).

```python
# À compléter
columns = df.columns

# Histogramme des âges
plt.figure(figsize=(4,6))
sns.histplot(df['age'])
plt.show()

# Distribution prix billet
plt.figure(figsize=(4,6))
sns.histplot(df['fare'], log_scale=True)
plt.show()

# Répartition des passager par classe
plt.figure(figsize=(4,6))
sns.countplot(df['class'])
plt.show()

# Répartition du nbre de survivant
plt.figure(figsize=(6,9))
sns.countplot(data = df, x='survived')
plt.show()
```

## 4. Relations entre variables

Étudiez maintenant les relations entre certaines variables :

1. Survie en fonction du sexe (`sns.countplot` avec `hue`).
2. Survie en fonction de la classe (`sns.countplot` avec `hue`).
3. Distribution des âges selon la survie (`sns.boxplot`).
4. Prix du billet selon la survie (`sns.violinplot`).

```python
# À compléter

# Survie en fonction du sexe
plt.figure(figsize=(4,6))
sns.countplot(data=df, x='sex',hue='survived')
plt.show()

# Survie en fonction de la classe
plt.figure(figsize=(4,6))
sns.countplot(data=df,x='class', hue='survived')
plt.show()

# distribution des ages selon la survie
plt.figure(figsize=(4,6))
sns.boxplot(data=df,x='age', hue='survived')
plt.show()

# Prix du billet selon la survie
plt.figure(figsize=(4,6))
sns.violinplot(data=df, x='fare',  hue='survived')
plt.show()
```

## 5. Analyses croisées

1. Comparez la **survie** selon le couple `(sex, class)` (`sns.catplot`).
2. Étudiez la relation entre `sibsp` (frères/sœurs/conjoints) et survie.
3. Étudiez la relation entre `parch` (parents/enfants) et survie.
4. Utilisez `pd.crosstab` pour résumer la survie par sexe et par classe.

```python
# À compléter

#Comparer la survie selon sex et class
plt.figure(figsize=(9,12))
sns.catplot(data=df,
            x='class',
            y='survived',
            hue='sex',
            kind='bar',
                )
plt.title("Taux de survie selon le sexe et la classe")
plt.ylabel("Probabilité de survie")
plt.xlabel("class")
plt.show()

# Etudier la relation entre sibsp et survie
plt.figure(figsize=(4,6))
sns.countplot(data=df,
              x='sibsp',
              hue='survived')
plt.title("Survie selon freres/soeurs/conjoint")
plt.show()

# Etudier la relation entre sibsp et survie
plt.figure(figsize=(4,6))
sns.countplot(data=df,
              x='parch',
              hue='survived')
plt.title("Survie selon liens de parentés")
plt.show()

# Crosstab
pd.crosstab([df['sex'], df['class']],
            df['survived']
)
```

## 6. Corrélations

1. Calculez la matrice de corrélation (`.corr()`).
2. Visualisez-la avec une heatmap (`sns.heatmap`).
3. Identifiez les variables les plus corrélées avec la survie.

```python
# À compléter
plt.figure(figsize=(8,8))
plt.title("heatmap de corrélation")
sns.heatmap(df.corr(numeric_only=True),cbar=False,annot=True,cmap='Blues',fmt='.2g')
plt.show()

# Les variables les plus corrélées avec la survie sont : ['Sexe','class']
```
