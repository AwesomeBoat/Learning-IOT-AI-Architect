## Pipeline

```python
df = sns.load_dataset("titanic")
display(df.head())
df = df[['age','sex','pclass','survived','alone']] # drop deck col

to_impute=['age']
to_encode=['sex']

impute_pipeline= Pipeline([("age_impute",SimpleImputer(strategy="mean"))])
encode_pipeline = Pipeline([("Sex_encode",OneHotEncoder())])

transformer = ColumnTransformer([("impute",impute_pipeline,to_impute),
                               ("encode",encode_pipeline,to_encode)], remainder='passthrough')


final_pipeline = Pipeline([("transform",transformer),
                           ('model',DecisionTreeClassifier())])

final_pipeline

```

```python
X = df.drop(columns='survived')
y = df['survived']

X_train, X_test, y_train, y_test = train_test_split(X,y,random_state=33,test_size=0.2)


final_pipeline.fit(X_train,y_train)

y_pred = final_pipeline.predict(X_test)
display(f"Accuracy score = {accuracy_score(y_test,y_pred)}")

#========================================

X_train, X_test, y_train, y_test = train_test_split(X,y,random_state=35,test_size=0.2)


final_pipeline.fit(X_train,y_train)

y_pred = final_pipeline.predict(X_test)
display(f"Accuracy score = {accuracy_score(y_test,y_pred)}")


#========================================
X_train, X_test, y_train, y_test = train_test_split(X,y,random_state=42,test_size=0.2)


final_pipeline.fit(X_train,y_train)

y_pred = final_pipeline.predict(X_test)
display(f"Accuracy score = {accuracy_score(y_test,y_pred)}")

#========================================
X_train, X_test, y_train, y_test = train_test_split(X,y,random_state=55,test_size=0.2)


final_pipeline.fit(X_train,y_train)

y_pred = final_pipeline.predict(X_test)
display(f"Accuracy score = {accuracy_score(y_test,y_pred)}")
```
